```python
from pathlib import Path
import numpy as np
import pandas as pd

# Definición de rutas relativas usando pathlib para asegurar reproducibilidad
BASE_DIR = Path("..")
DATA_RAW = BASE_DIR / "data" / "raw"
DATA_PROCESSED = BASE_DIR / "data" / "processed"
OUTPUTS = BASE_DIR / "outputs"

# Crear las carpetas si no existen
DATA_RAW.mkdir(parents=True, exist_ok=True)
DATA_PROCESSED.mkdir(parents=True, exist_ok=True)
OUTPUTS.mkdir(parents=True, exist_ok=True)

print("Entorno virtual y rutas relativas configuradas correctamente.")
```

```python
# 1. Lista simple de aprendices
aprendices = ["Ana", "Carlos", "María", "Luis", "Sofía"]
print("Lista de aprendices:", aprendices)
print("Cantidad total:", len(aprendices))

# 2. Lista de diccionarios (con los 2 registros adicionales requeridos)
datos_aprendices = [
    {
        "documento": "1001",
        "nombre": "Ana",
        "programa": "ADSO",
        "nota_python": 4.2,
        "asistencia": 92,
    },
    {
        "documento": "1002",
        "nombre": "Carlos",
        "programa": "ADSO",
        "nota_python": 3.8,
        "asistencia": 85,
    },
    {
        "documento": "1003",
        "nombre": "María",
        "programa": "ADSO",
        "nota_python": 4.7,
        "asistencia": 96,
    },
    # Reto resuelto: 2 nuevos registros agregados
    {
        "documento": "1004",
        "nombre": "Luis",
        "programa": "ADSO",
        "nota_python": 2.9,
        "asistencia": 70,
    },
    {
        "documento": "1005",
        "nombre": "Sofía",
        "programa": "ADSO",
        "nota_python": 5.0,
        "asistencia": 98,
    },
]

print("\nLista de diccionarios estructurada con", len(datos_aprendices), "registros.")
```

```python
# Arreglo NumPy con las notas de Python
notas = np.array([4.2, 3.8, 4.7, 2.9, 5.0])

print("--- Estadísticas básicas con NumPy ---")
print("Notas:", notas)
print("Promedio:", np.mean(notas))
print("Mediana:", np.median(notas))
print("Nota Máxima:", np.max(notas))
print("Nota Mínima:", np.min(notas))

# Operación vectorizada: bonificación de 0.2 sin superar 5.0
notas_bonificadas = np.clip(notas + 0.2, 0, 5.0)
print("\nNotas Bonificadas (+0.2):", notas_bonificadas)

# Filtro numérico
notas_aprobadas = notas[notas >= 3.5]
porcentaje_aprobacion = (len(notas_aprobadas) / len(notas)) * 100
print(f"\nPorcentaje de aprobación (>= 3.5): {porcentaje_aprobacion:.1f}%")
```

```python
# Crear una Serie de Pandas con índices personalizados
serie_notas = pd.Series(
    data=[4.2, 3.8, 4.7, 2.9, 5.0],
    index=["Ana", "Carlos", "María", "Luis", "Sofía"],
)
print("--- Serie de Pandas ---")
print(serie_notas)

# Crear DataFrame a partir de la lista de diccionarios
df_aprendices = pd.DataFrame(datos_aprendices)

# Agregar columna calculada "estado" con np.where
df_aprendices["estado"] = np.where(
    df_aprendices["nota_python"] >= 3.5, "Aprobado", "En riesgo"
)

# Reto resuelto: Columna 'alerta_asistencia'
df_aprendices["alerta_asistencia"] = np.where(
    df_aprendices["asistencia"] < 80, "Revisar asistencia", "Sin alerta"
)

df_aprendices
```

```python
# Dataset base ampliado
datos_base = {
    "documento": ["1001", "1002", "1003", "1004", "1005"],
    "nombre": ["Ana", "Carlos", "María", "Luis", "Sofía"],
    "programa": ["ADSO", "ADSO", "ADSO", "ADSO", "ADSO"],
    "ficha": ["287001", "287001", "287001", "287001", "287001"],
    "nota_python": [4.2, 3.8, 4.7, 2.9, 5.0],
    "nota_datos": [4.0, 3.6, 4.8, 3.1, 4.9],
    "asistencia": [92, 85, 96, 70, 98],
    "ciudad": ["Medellín", "Bello", "Itagüí", "Medellín", "Envigado"],
}
df_base = pd.DataFrame(datos_base)

# Definición de rutas relativas
ruta_csv = DATA_RAW / "seguimiento_aprendices_semana2.csv"
ruta_excel = DATA_RAW / "seguimiento_aprendices_semana2.xlsx"

# Exportar a raw para simular fuentes externas
df_base.to_csv(ruta_csv, index=False, encoding="utf-8")
df_base.to_excel(ruta_excel, index=False)

# Lectura de los archivos usando Pandas
df_csv = pd.read_csv(ruta_csv)
df_excel = pd.read_excel(ruta_excel)

print(f"CSV cargado correctamente. Forma (shape): {df_csv.shape}")
print(f"Excel cargado correctamente. Forma (shape): {df_excel.shape}")
```

```python
df_seguimiento = pd.read_csv(ruta_csv)

print("--- Primeros 3 registros (head) ---")
display(df_seguimiento.head(3))

print("--- Últimos 2 registros (tail) ---")
display(df_seguimiento.tail(2))

print("--- Información del DataFrame (info) ---")
df_seguimiento.info()

print("\n--- Estadísticas descriptivas (describe) ---")
display(df_seguimiento.describe())
```

## 🔍 Hallazgos Iniciales del Dataset
- **Dimensiones:** El dataset cuenta con **5 filas y 8 columnas**.
- **Variables Numéricas:** `nota_python`, `nota_datos`, `asistencia`.
- **Variables Categóricas / Texto:** `documento`, `nombre`, `programa`, `ficha`, `ciudad`.
- **Análisis de Rendimiento:** 
  - El promedio de `nota_python` es **4.12** y el de `nota_datos` es **4.08**.
  - El aprendiz **Luis (1004)** presenta una nota en Python por debajo de 3.5 (2.9) y un porcentaje de asistencia del 70%, por lo que requiere acompañamiento.
- **Atención para Semana 3:** Revisar límites de asistencia y alertas tempranas para prevención de deserción.

```python
# Calcular promedio general y estado académico
df_seguimiento["promedio_general"] = (
    df_seguimiento["nota_python"] + df_seguimiento["nota_datos"]
) / 2
df_seguimiento["estado_academico"] = np.where(
    df_seguimiento["promedio_general"] >= 3.5, "Aprobado", "En riesgo"
)

# Filtrar y ordenar columnas finales
df_final = df_seguimiento[[
    "documento",
    "nombre",
    "programa",
    "ficha",
    "nota_python",
    "nota_datos",
    "promedio_general",
    "asistencia",
    "estado_academico",
    "ciudad",
]]

# Rutas de salida procesadas
ruta_csv_processed = DATA_PROCESSED / "seguimiento_semana2_procesado.csv"
ruta_excel_processed = DATA_PROCESSED / "seguimiento_semana2_procesado.xlsx"

# Exportar
df_final.to_csv(ruta_csv_processed, index=False, encoding="utf-8")
df_final.to_excel(ruta_excel_processed, index=False)

print("Archivos exportados exitosamente a data/processed/:")
print(" -", ruta_csv_processed)
print(" -", ruta_excel_processed)

df_final
```

## 📌 Conclusiones Técnicas de la Semana 2

1. **Listas vs. Diccionarios vs. DataFrames:**
   - Una **lista** es una secuencia ordenada de elementos (ideal para listas simples de valores o nombres).
   - Un **diccionario** guarda pares clave-valor (ideal para representar las propiedades de una única entidad o registro).
   - Un **DataFrame** de Pandas organiza los datos en formato matricial (filas y columnas), permitiendo operaciones vectorizadas y análisis estadísticos a gran escala.

2. **Diferencia entre `info()` y `describe()`:**
   - `info()` muestra la estructura técnica del DataFrame: número de entradas, nombres de columnas, cantidad de valores no nulos y tipos de datos (`dtypes`).
   - `describe()` genera un resumen estadístico descriptivo (media, desviación estándar, min, max, cuartiles) para variables numéricas.

3. **Uso de Rutas Relativas (`pathlib.Path`):**
   - Garantiza que el código y el notebook sean totalmente **reproducibles** en cualquier sistema operativo (Windows, Linux, macOS) o equipo sin depender de rutas absolutas locales (`C:\Users\...`).
